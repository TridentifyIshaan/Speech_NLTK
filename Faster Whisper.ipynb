## Importing Packages

```python
from faster_whisper import WhisperModel
```

## Load faster-whisper model
- Options: tiny, base, small, medium, large-v2, large-v3
- device: "cpu" or "cuda"
- compute_type: "int8", "float16", "float32"

```python
model = WhisperModel("small", device="cpu", compute_type="int8")
```

## Setting Audio Path

```python
audio_path = "French3.mp3"
```

## Transcribe and translate from French to English

```python
segments, info = model.transcribe(
    audio_path,
    task="translate",      # "translate" to English or "transcribe" to keep original language
    language="fr",         # source language
    beam_size=5
)

print(f"Detected language '{info.language}' with probability {info.language_probability}")
```

## Collect all segments into a single translation

```python
translation = " ".join([segment.text for segment in segments])
print("Translated text:", translation)
```
